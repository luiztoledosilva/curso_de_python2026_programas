```python
def raiz_quadrada(x):
    if x < 0:
        return "Não é possível calcular a raiz quadrada de um número negativo."
    elif x == 0:
        return 0
    else:
        return x ** 0.5
print(raiz_quadrada(16))  # Deve retornar 4.0
print(raiz_quadrada(-4))  # Deve retornar a mensagem de erro                                
```
