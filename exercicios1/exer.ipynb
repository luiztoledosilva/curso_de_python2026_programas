```python
primeiro_nome="Luiz"
ultimo_nome="Silva"
nome_completo=primeiro_nome+" "+ultimo_nome
print(nome_completo)
```

```python
###inverter a ordem de uma string
nome_completo_invertido=nome_completo[::-1]
print(nome_completo_invertido)              
```
